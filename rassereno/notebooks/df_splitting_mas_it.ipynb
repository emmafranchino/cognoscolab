```python
gpt_4o_demographics = pd.read_csv('/Users/emma/projects/cognoscolab/math_anxiety_LLMs/out/formatting/gpt_4o/all_questionnaires_gpt_4o.csv')
gpt_4o_fisica_participants = gpt_4o_demographics['participant_id'][gpt_4o_demographics['faculty'].astype(str) == 'uni_fisica']
gpt_4o_psicologia_participants = gpt_4o_demographics['participant_id'][gpt_4o_demographics['faculty'].astype(str) == 'uni_psicologia']
gpt_4o_liceo_participants = gpt_4o_demographics['participant_id'][gpt_4o_demographics['faculty'].str.startswith('liceo')]
len(gpt_4o_liceo_participants) + len(gpt_4o_fisica_participants) + len(gpt_4o_psicologia_participants)
```

```python
gpt_4o_fisica_ass_val = association_valence_data_gpt_4o[association_valence_data_gpt_4o['participant_id'].isin(gpt_4o_fisica_participants)]
gpt_4o_fisica_mas_it = mas_it_gpt_4o[mas_it_gpt_4o['participant_id'].isin(gpt_4o_fisica_participants)]
gpt_4o_fisica_ass_val.to_csv('/Users/emma/projects/cognoscolab/rassereno/out/formatting/fisica_gpt-4o.csv')

with open(f'{rassereno_path}/out/formatting/participants_rassereno.txt', 'a') as file:
        file.write("\n"
                   f"physics_students_gpt-4o: {len(gpt_4o_fisica_mas_it)}")
```

```python
gpt_4o_psicologia_ass_val = association_valence_data_gpt_4o[association_valence_data_gpt_4o['participant_id'].isin(gpt_4o_psicologia_participants)]
gpt_4o_psicologia_mas_it = mas_it_gpt_4o[mas_it_gpt_4o['participant_id'].isin(gpt_4o_psicologia_participants)]
low_high_anxiety_division(gpt_4o_psicologia_mas_it, gpt_4o_psicologia_ass_val, 'psychology_students_gpt_4o', rassereno_path)
```
